### Carga de datos

```python
import os

# Tu competencia de Kaggle
COMPETITION = "hubmap-hacking-the-human-vasculature"

# Verificar espacio disponible en Colab
print("Espacio disponible en Colab:")
!df -h /content

# Descargar datos de la competencia
print(f"\nDescargando datos de {COMPETITION}...")
print("   Esto tomará ~5-10 minutos para 4.5GB\n")

# Crear directorio para datos
!mkdir -p /content/hubmap_data
%cd /content/hubmap_data

# Descargar (necesitas haber aceptado las reglas de la competencia en Kaggle)
!kaggle competitions download -c {COMPETITION}

print("\nDescarga completada")

# Ver archivos descargados
print("\nArchivos descargados:")
!ls -lh
```

### Descomprimir datos

```python
import zipfile
from tqdm import tqdm

print("Descomprimiendo archivos...")

# Descomprimir el archivo principal
zip_file = f"{COMPETITION}.zip"

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    # Obtener lista de archivos
    file_list = zip_ref.namelist()

    # Descomprimir con barra de progreso
    for file in tqdm(file_list, desc="Extrayendo"):
        zip_ref.extract(file, '/content/hubmap_data/')

print("\nDatos descomprimidos")

# Listar estructura
print("\nEstructura de datos:")
!ls -lh /content/hubmap_data/

# Verificar que train/ existe
if os.path.exists('/content/hubmap_data/train'):
    num_images = len(os.listdir('/content/hubmap_data/train'))
    print(f"\nCarpeta train/ encontrada con {num_images} imágenes")
else:
    print("\nRevisando estructura...")
    !find /content/hubmap_data -name "train" -type d
```
